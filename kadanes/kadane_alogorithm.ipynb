```python
# https://leetcode.com/problems/maximum-subarray/description/
# 53. Maximum Subarray

"""
Input: nums = [-2,1,-3,4,-1,2,1,-5,4]
Output: 6
Explanation: The subarray [4,-1,2,1] has the largest sum 6.
"""

nums = [-2,1,-3,4,-1,2,1,-5,4]
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        max_sum = float("-inf")
        current_max = 0

        for index in range(len(nums)):
            current_max = max(nums[index], current_max+nums[index])
            max_sum = max(max_sum, current_max)
        return max_sum
        

        
        
```

```python
### 152. Maximum Product Subarray
### https://leetcode.com/problems/maximum-product-subarray/description/
"""
Example 1:

Input: nums = [2,3,-2,4]
Output: 6
Explanation: [2,3] has the largest product 6.
Example 2:

Input: nums = [-2,0,-1]
Output: 0
Explanation: The result cannot be 2, because [-2,-1] is not a subarray.
"""
class Solution:
    def maxProduct(self, nums: List[int]) -> int:

        current_max_pro = nums[0]
        current_min_pro = nums[0]

        max_product = nums[0]

        for val in nums[1:]:

            if val < 0:
                current_max_pro, current_min_pro = (
                    current_min_pro,
                    current_max_pro
                )

            current_max_pro = max(val, current_max_pro * val)

            current_min_pro = min(val, current_min_pro * val)

            max_product = max(max_product, current_max_pro)

        return max_product
```

```python
### 918. Maximum Sum Circular Subarray

"""
Intuition in One Sentence

Maximum circular subarray =

either normal Kadane maximum
or total array minus the minimum middle subarray


answer = max(
    normal_max_subarray,
    total_sum - min_subarray
)
"""


"""
Important Edge Case

Suppose:

nums = [-3, -2, -5]

Then:

total_sum = -10
min_subarray = -10

So:

total_sum - min_subarray = 0

But 0 is invalid because we cannot take an empty subarray.

Correct answer is:

-2

So:

If all numbers are negative:

return max_sum
"""
class Solution:
    def maxSubarraySumCircular(self, nums: List[int]) -> int:

        total_sum = sum(nums)

        current_max = 0
        max_sum = nums[0]

        current_min = 0
        min_sum = nums[0]

        for val in nums:

            # Kadane for maximum subarray
            current_max = max(val, current_max + val)
            max_sum = max(max_sum, current_max)

            # Kadane for minimum subarray
            current_min = min(val, current_min + val)
            min_sum = min(min_sum, current_min)

        # all numbers negative
        if max_sum < 0:
            return max_sum

        return max(max_sum, total_sum - min_sum)
```
