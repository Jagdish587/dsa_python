```python
### https://leetcode.com/problems/linked-list-cycle-ii/description/
### 142. Linked List Cycle II

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # find if loop exists

            slow = head
            fast = head

            while fast and fast.next != None:
                slow = slow.next
                fast = fast.next.next

                if slow == fast:
                   break
            else:
                return None

            slow = head
            while slow != fast:
                slow = slow.next
                fast = fast.next

            return slow     

        
```

```python
### https://leetcode.com/problems/happy-number/description/
### 202. Happy Number
class Solution:
    def isHappy(self, n: int) -> bool:
        def get_next(n):
            total = 0
            while n:
                digit = n % 10
                n = n // 10
                total += digit ** 2
            return total
        
        slow = n
        fast = get_next(n)

        while fast != 1 and slow != fast:
            slow = get_next(slow)
            fast = get_next(get_next(fast))
        
        return fast == 1
```

```python
### https://leetcode.com/problems/palindrome-linked-list/description/
### 234. Palindrome Linked List

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def isPalindrome(self, head: ListNode | None) -> bool:
        current_node = head
        mystack  = []
        while current_node:
            mystack.append(current_node.val)
            current_node = current_node.next

        current_node = head
        while current_node:
            val = mystack.pop()
            if current_node.val != val:
                return False 
            current_node = current_node.next
        return True
        
```

```python
### https://leetcode.com/problems/reverse-linked-list/description/
### 206. Reverse Linked List


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverseList(self, head: ListNode | None) -> ListNode | None:
        current_node = head
        prev_node = None
        next_node =  None

        while current_node:
            next_node = current_node.next
            current_node.next =  prev_node
            prev_node = current_node
            current_node = next_node

        return prev_node
```

```python
### https://leetcode.com/problems/intersection-of-two-linked-lists/
### 160. Intersection of Two Linked Lists

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, x):
#         self.val = x
#         self.next = None

class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        myset = set()
        current_node_A = headA
        current_node_B = headB

        while current_node_A != None:
            myset.add(current_node_A)
            current_node_A = current_node_A.next
        
        while current_node_B != None:
            if current_node_B in myset:
                return current_node_B
            current_node_B = current_node_B.next
        else:
            return None

```

```python
### https://leetcode.com/problems/remove-duplicates-from-sorted-list-ii/description/
### 82. Remove Duplicates from Sorted List II


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def deleteDuplicates(self, head: ListNode | None) -> ListNode | None:
        dummy_node = ListNode(-1)
        dummy_node.next = head

        prev_node = dummy_node
        current_node = head

        current_node = head
        while current_node:
            if current_node.next and current_node.val == current_node.next.val:
                duplicate_val = current_node.val
                while current_node and duplicate_val == current_node.val:
                    current_node = current_node.next 
                prev_node.next = current_node
            else:    
                prev_node = current_node
                current_node = current_node.next
        return dummy_node.next
```

```python
### https://leetcode.com/problems/remove-nth-node-from-end-of-list/description/
### 19. Remove Nth Node From End of List

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: ListNode | None, n: int) -> ListNode | None:
        dummy_node = ListNode(-1)
        dummy_node.next = head


        right_node = dummy_node
        for _ in range(n+1):
            right_node = right_node.next

        left_node = dummy_node
        while right_node:
            right_node =  right_node.next
            left_node = left_node.next
        left_node.next = left_node.next.next
        
        return dummy_node.next
```

```python
### https://leetcode.com/problems/swap-nodes-in-pairs/description/
### 24. Swap Nodes in Pairs
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def swapPairs(self, head: ListNode | None) -> ListNode | None:
        if head is None:
            return head
        dummy_node = ListNode(-1)
        dummy_node.next = head
        head = dummy_node
        prev_node = dummy_node
        left_node = head.next
        right_node = left_node.next


        while left_node.next:       
            prev_node.next = right_node
            left_node.next = right_node.next
            right_node.next = left_node
            if left_node.next:
                prev_node = left_node
                left_node = left_node.next
                right_node = left_node.next
        
        head = dummy_node.next
        return head
```
