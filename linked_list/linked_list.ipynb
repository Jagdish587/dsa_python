```python
### https://leetcode.com/problems/lfu-cache/description/
### 460. LFU Cache

"""
lru_list	DoubleLinkedList
lru_list.left_dummy_node	Node
lru_list.left_dummy_node.next	Node
lru_list.left_dummy_node.next.key	int (assuming integer keys
"""

from collections import defaultdict
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.freq = 1
        self.prev = None
        self.next = None

class DoubleLinkedList:

    def __init__(self):
        self.left_dummy_node = Node(0,0)
        self.right_dummy_node = Node(0,0)
        self.left_dummy_node.next = self.right_dummy_node
        self.right_dummy_node.prev = self.left_dummy_node
        self.size = 0


    def remove_node(self, node):
        prev_node = node.prev
        next_node = node.next
        prev_node.next = next_node
        next_node.prev = prev_node
        self.size -= 1 

    def insert_node(self, node):
        prev_node = self.right_dummy_node.prev
        next_node = self.right_dummy_node
        node.next = next_node
        node.prev = prev_node
        prev_node.next = node
        next_node.prev = node
        self.size += 1

class LFUCache:

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {} # key -> Node
        self.freq_map_list = defaultdict(DoubleLinkedList) # freq -> doubled linked list
        self.min_freq = 0

    def increment_freq(self, node):
        freq = node.freq
        self.freq_map_list[freq].remove_node(node)
        if self.freq_map_list[freq].size == 0:
            del self.freq_map_list[freq]

            if self.min_freq == freq:
                self.min_freq += 1

        node.freq += 1
        self.freq_map_list[node.freq].insert_node(node)

    def get(self, key: int) -> int:
        if key in self.cache:           
            self.increment_freq(self.cache[key])
            return self.cache[key].value
        else:
            return -1

    def put(self, key: int, value: int) -> None:
        if self.capacity == 0:
            return

        if key in self.cache:
            node = self.cache[key]
            node.value = value
            self.increment_freq(node)
            return

        if len(self.cache) >= self.capacity:
            lru_list = self.freq_map_list[self.min_freq]
            lru_node = lru_list.left_dummy_node.next
            lru_list.remove_node(lru_node)
            del self.cache[lru_node.key]
            if lru_list.size == 0:
                del self.freq_map_list[self.min_freq]

        node = Node(key, value)
        self.cache[key] = node
        self.min_freq = 1
        self.freq_map_list[1].insert_node(node)

# Your LFUCache object will be instantiated and called as such:
# obj = LFUCache(capacity)
# param_1 = obj.get(key)
# obj.put(key,value)
```

```python
### https://leetcode.com/problems/design-circular-queue/description/
### 622. Design Circular Queue

class Node:
    def __init__(self, val):
        self.val = val
        self.next = None

class MyCircularQueue:

    def __init__(self, k: int):
        self.max_size = k
        self.current_size = 0
        self.head = None
        self.tail = None

    def enQueue(self, value: int) -> bool: # push element to rear
        if self.current_size < self.max_size:
            new_node = Node(value)
            # insert element in queue
            if self.head == None:
                new_node.next = new_node
                self.head = new_node
                self.tail = new_node
            else:
                new_node.next = self.head
                self.tail.next = new_node
                self.tail = new_node
            self.current_size += 1
            return True
        else:
            return False
        

    def deQueue(self) -> bool: # pop element from front
        if self.current_size > 0:
            if self.head == self.tail:
                self.head = None
                self.tail = None
            else:
                self.tail.next = self.head.next
                self.head = self.head.next
            self.current_size -= 1
            return True
        else:
            return False
        

    def Front(self) -> int:
        if self.head:
            return self.head.val
        else: 
            return -1
        

    def Rear(self) -> int:
        if self.tail:
            return self.tail.val
        else:
            return -1
        

    def isEmpty(self) -> bool:
        if self.current_size == 0:
            return True
        else:
            return False
        

    def isFull(self) -> bool:
        if self.current_size == self.max_size:
            return True
        else:
            return False
```

```python
### https://leetcode.com/problems/reverse-nodes-in-k-group/description/
### 25. Reverse Nodes in k-Group

# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next

def reverse_nodes(left_node, right_node):
    prev_node =  None
    current_node = left_node


    while current_node != right_node:
        next_node = current_node.next
        current_node.next = prev_node
        prev_node = current_node
        current_node = next_node

    return prev_node


class Solution:
    def reverseKGroup(self, head: ListNode | None, k: int) -> ListNode | None:
        dummy_node = ListNode(-1)
        dummy_node.next = head
        left_node = head
        right_node = dummy_node
        prev_left = dummy_node

        while (1):
            for _ in range(k):
                if right_node:
                    right_node = right_node.next
            if right_node:
                next_node = right_node.next
                reversed_head = reverse_nodes(left_node, next_node)
                prev_left.next = reversed_head
                left_node.next = next_node
                prev_left = left_node
                right_node = left_node
                left_node = next_node
            else:
                break
            
        return dummy_node.next
```

```python
### https://leetcode.com/problems/rotate-list/description/
### 61. Rotate List

from collections import deque
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def rotateRight(self, head: ListNode | None, k: int) -> ListNode | None:
        current_node = head
        new_list = deque()
        while current_node != None:
            new_list.append(current_node)
            current_node = current_node.next

        new_list.rotate(k)
        dummy_node = ListNode(-1)
        new_head = dummy_node

        while new_list:
            current_node = new_list.popleft()
            dummy_node.next = current_node
            dummy_node = current_node
            dummy_node.next = None

        head = new_head.next
        return head
```

```python
### https://leetcode.com/problems/partition-list/description/
### 86. Partition List


# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def partition(self, head: ListNode | None, x: int) -> ListNode | None:
        current_node = head
        
        dummy_node = ListNode(-1)
        smaller_value_list_head = dummy_node
        smaller_value_list = smaller_value_list_head

        dummy_node_2 = ListNode(-2)
        larger_value_list_head = dummy_node_2
        larger_value_list = larger_value_list_head

        while current_node != None:
            new_node = ListNode(current_node.val)
            if current_node.val < x:
                smaller_value_list.next = new_node
                smaller_value_list = new_node
            else:
                larger_value_list.next = new_node
                larger_value_list = new_node
                larger_value_list.next = None
            current_node = current_node.next
    
        smaller_value_list.next = larger_value_list_head.next
        head = smaller_value_list_head.next
        return head
```

```python
### https://leetcode.com/problems/flatten-a-multilevel-doubly-linked-list/description/
### 430. Flatten a Multilevel Doubly Linked List

"""
# Definition for a Node.
class Node:
    def __init__(self, val, prev, next, child):
        self.val = val
        self.prev = prev
        self.next = next
        self.child = child
"""

class Solution:
    def flatten(self, head: 'Optional[Node]') -> 'Optional[Node]':
        current_node = head
        mystack = []


        while current_node:
            if current_node.child:
                if current_node.next:
                    mystack.append(current_node.next)

                current_node.next = current_node.child
                current_node.child.prev = current_node
                current_node.child = None
            

            if not current_node.next and mystack:
                next_node = mystack.pop()
                current_node.next = next_node
                next_node.prev = current_node

            current_node = current_node.next
        return head
```
