```python
### 974. Subarray Sums Divisible by K

### Main Point : If two prefix sums have the same remainder when divided by 𝐾 ,
### the subarray between these two prefix sums is divisible by 𝐾
### If the SAME remainder appears again, the subarray between them is divisible by k

### same remainder at i and j → subarray (i+1 to j) 
### is divisible by k
from collections import defaultdict
nums = [4,5,0,-2,-3,1]
k = 5

seen = defaultdict(int)
seen[0] = 1
prefix = 0
count = 0
for num in nums:
    rem = (prefix + num) % k
    count += seen[rem]
    seen[rem] += 1
    print("prefix = ", prefix, "seen = ", seen, "count = ", count)

print("count = ", count)
```

```python
### 325. Maximum Size Subarray Sum Equals k
def maxSubArrayLen(nums, k):
    prefix = {0: -1}

    curr_sum = 0
    max_len = 0

    for right_index, num in enumerate(nums):
        curr_sum += num

        need = curr_sum - k

        if need in prefix:
            max_len = max(
                max_len,
                right_index - prefix[need]
            )

        # Store FIRST occurrence only
        if curr_sum not in prefix:
            prefix[curr_sum] = right_index

    return max_len
```

```python
### https://leetcode.com/problems/range-sum-query-immutable/description/
### 303. Range Sum Query - Immutable

class NumArray:

    def __init__(self, nums: List[int]):
        self.res = [0] * len(nums)
        self.res[0] = nums[0]

        for index in range(1, len(nums)):
            self.res[index] = self.res[index-1] + nums[index]
        

    def sumRange(self, left: int, right: int) -> int:
        sumRange = 0
        if left == 0:
            sumRange = self.res[right]
        else:
            sumRange = self.res[right] - self.res[left-1]
        return sumRange
        


# Your NumArray object will be instantiated and called as such:
# obj = NumArray(nums)
# param_1 = obj.sumRange(left,right)
```

```python
### https://leetcode.com/problems/subarray-sum-equals-k/description/
### 560. Subarray Sum Equals K
from collections import defaultdict
nums = [2,-1,1,2]
k = 2
prefix_count = defaultdict(int)
prefix_count[0] = 1

current_sum = 0
count = 0

for num in nums:
    current_sum += num

    # Count previous prefix sums that would form sum k
    count += prefix_count.get(current_sum - k, 0)

    # Record current prefix sum
    prefix_count[current_sum] += 1

print("count = ", count)

"""
Intution:
For every current prefix sum:

find how many previous prefix sums equal curr-k
add that frequency to answer

That’s literally the entire trick.
"""
```

```python
### https://leetcode.com/problems/continuous-subarray-sum/description/
### 523. Continuous Subarray Sum

### Take reminder index map
class Solution:
    def checkSubarraySum(self, nums: List[int], k: int) -> bool:
        reminder_index = {0: -1} # This allows us to detect subarrays that start at index 0.
        current_sum = 0
        rem = 0

        for index in range(len(nums)):
            current_sum += nums[index]
            rem = current_sum % k
            if rem in reminder_index:
                lenght = index - reminder_index[rem]
                if lenght >= 2:
                    return True
            else:
                reminder_index[rem] = index
        return False
```

```python
### https://leetcode.com/problems/contiguous-array/description/
### 525. Contiguous Array
class Solution:
    def findMaxLength(self, nums: List[int]) -> int:
        seen = {0:-1}
        max_length = 0
        current_sum = 0

        for index,value in enumerate(nums):
            if value == 0:
                value = -1
            current_sum += value

            if current_sum in seen:
                length = index - seen[current_sum]
                max_length = max(max_length, length)
            else:
                seen[current_sum] = index
        return max_length        
```

```python

#### lets say this is input [1, -5, 2, -3, -7, 8, 6] 
### return True if any of sub array sum equal to 0 else False

def has_zero_sum_subarray(arr):
    seen = {0}
    prefix_sum = 0

    for num in arr:
        prefix_sum += num

        if prefix_sum in seen:
            return True

        seen.add(prefix_sum)

    return False


arr = [1, -5, 2, -3, -7, 8, 6]
print(has_zero_sum_subarray(arr)) 
```
