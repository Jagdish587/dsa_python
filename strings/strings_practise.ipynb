```python
# Count uppercase, lowercase, digits, and special characters.
mystr = "!PythLonN@2025?"
upper_case_count = sum(1 for ch in mystr if ch.isupper())
lower_case_count = sum(1 for ch in mystr if ch.islower())
digit_count = sum(1 for ch in mystr if ch.isdigit())
special_symbols_count = sum(1 for ch in mystr if not ch.isalnum())

print("upper case count = ", upper_case_count)
print("lower case count = ", lower_case_count)
print("digit count = ", digit_count)
print("special symbols count = ", special_symbols_count)
```

```python
# Find duplicate characters in a string.
mystr = "mississippi"
unqiue_str = set()
print("unqiue_str = ", unqiue_str)

duplicate_ch = ""
for ch in mystr:
    if ch not in unqiue_str:
        unqiue_str.add(ch)
    else:
        duplicate_ch += ch

str_form = "".join(sorted(set(duplicate_ch)))
print("duplicate_ch = ", str_form)
```

```python
# Find the first non-repeated character
mystr = "programming"
for ch in mystr:
    
    if mystr.count(ch) == 1:
        print(ch)
        break
```

```python
# Remove duplicate characters while preserving order
mystr = "mississippi"
res = ""
for ch in mystr:
    if ch not in res:
        res += ch
print("res = ", len(res))
```

```python
# Count frequency of each character.
from collections import Counter
mystr = "mississippi"
freq_counter = Counter(mystr)
print(freq_counter)


from collections import defaultdict
mystr = "mississippi"

freq_counter = defaultdict(int)

for ch in mystr:
    freq_counter[ch] = freq_counter.get(ch, 0) + 1

print("freq_counter = ", freq_counter)
```

```python
# Find the longest word in a sentence
mystr = "I love Python programming"
res_list = mystr.split()

word = res_list[0]
for value in res_list[1:]:
    if len(value) > len(word):
        word = value

print("word = ", word)
```

```python
# Compress a string.
mystr = "aaabbcc"
char_freq = Counter(mystr)

res = ""
for key,value in char_freq.items():
    res += key + str(value)

print("res = ", res)
```

```python
# Find all substrings of a string
mystr = "abc"

for i in range(len(mystr)):
    for j in range(i + 1, len(mystr) + 1):
        print(mystr[i:j])
```

```python
# Find all permutations of a string
from itertools import permutations

mystr = "ABC"
for val in permutations(mystr):
    print("".join(val))

```

```python
# Find the most frequent character.
mystr = "mississippi"
print(Counter(mystr).most_common(1)[0][0])
```

```python
# Count occurrences of a substring without using count().
mystr = "abababa"
sub = "aba"

count = 0
start = 0

while True:
    pos = mystr.find(sub, start)

    if pos == -1:
        break

    count += 1
    start = pos + 1      # Move by 1 to count overlapping occurrences

print(count)
```
