```python

Monotonic Increasing Stack

Elements are arranged from smallest at the bottom to largest at the top.

Example: [1, 3, 5, 8]


Monotonic Decreasing Stack

Elements are arranged from largest at the bottom to smallest at the top.

Example: [8, 5, 3, 1]
```

```python
### design monotonic increasing stack o/p = [1,2,4]
nums = [1, 3, 2, 4]
my_stack = []


for val in nums:

    while my_stack and val < my_stack[-1]:
        my_stack.pop()

    my_stack.append(val)

print("my stack = ", my_stack) 
```

```python
# design monotonic decreasing stack o/p = [3, 2, 1]
nums = [1, 3, 2, 1]
my_stack = []

for val in nums:

    while my_stack and val > my_stack[-1]:
        my_stack.pop()
    my_stack.append(val)

print("my stack = ", my_stack)
```

```python
### https://leetcode.com/problems/next-greater-element-i/description/
### 496. Next Greater Element I
class Solution:
    def nextGreaterElement(self, nums1: list[int], nums2: list[int]) -> list[int]:
        next_greater  = {}
        my_stack = []
        res = []
        for num in nums2:
            while my_stack and num > my_stack[-1]:
                smaller_val = my_stack.pop()
                next_greater[smaller_val] = num
            my_stack.append(num)
        
        for val in my_stack:
            next_greater[val] = -1

        for num in nums1:
            res.append(next_greater[num])
        
        return res
```

```python
### https://leetcode.com/problems/daily-temperatures/description/   
### 739. Daily Temperatures
class Solution:
    def dailyTemperatures(self, temperatures: list[int]) -> list[int]:
        my_stack = []
        res = [0] * len(temperatures)
        for current_index, num in enumerate(temperatures):
            while my_stack and num > my_stack[-1][0]:
                past_temp, past_index = my_stack.pop()
                res[past_index] = current_index - past_index
            my_stack.append((num, current_index))
        return res
```

```python
### https://leetcode.com/problems/online-stock-span/description/
### 901. Online Stock Span

class StockSpanner:

    def __init__(self):
        self.my_stack = []
        

    def next(self, price: int) -> int:
        span = 1
        while self.my_stack and price >= self.my_stack[-1][0]:

            previous_price, previous_span = self.my_stack.pop()
            span += previous_span

        self.my_stack.append((price, span))
        return span
```

```python
### https://leetcode.com/problems/132-pattern/description/
### 456. 132 Pattern
class Solution:
    def find132pattern(self, nums: list[int]) -> bool:
        
        second_element = float("-inf")
        my_stack = []
        for num in reversed(nums):

            if num < second_element:
                return True

            while my_stack and num >  my_stack[-1]:
                second_element = my_stack.pop()    
    
            my_stack.append(num)
        
        return False
```

```python
### https://leetcode.com/problems/number-of-visible-people-in-a-queue/description/
### 1944. Number of Visible People in a Queue
class Solution:
    def canSeePersonsCount(self, heights: list[int]) -> list[int]:
        res = [0] * len(heights)
        my_stack = []
        count = 0

        for index in range(len(heights)-1, -1, -1):
            count = 0
            while my_stack and heights[index] > my_stack[-1]:
                my_stack.pop()
                count +=1 
            
            if my_stack:
                count +=1
            
            res[index] = count
            my_stack.append(heights[index])
        return res
```

```python
### https://leetcode.com/problems/largest-rectangle-in-histogram/description/
### 84. Largest Rectangle in Histogram
class Solution:
    def largestRectangleArea(self, heights: list[int]) -> int:
        
        my_stack = []
        max_area = 0

        for current_index, current_height in enumerate(heights):
            start_index = current_index

            while my_stack and current_height < my_stack[-1][-1]:
                past_index, past_height = my_stack.pop()
                area = (current_index - past_index) * past_height # width * height
                max_area = max(max_area, area)
                start_index = past_index

            my_stack.append((start_index, current_height))

        for index, height in my_stack:
            area = (len(heights) - index) * height
            max_area = max(max_area, area)
        

        return max_area
```
