```python
### https://leetcode.com/problems/valid-parentheses/
### 20. Valid Parentheses

class Solution:
    def isValid(self, s: str) -> bool:
        mystack = []

        for ch in s:
            if ch == "(" or ch == "[" or ch == "{":
                mystack.append(ch)
            else:
                if ch == ")":
                    if mystack and mystack[-1] == "(":
                        mystack.pop()
                    else:
                        return False
                    
                elif ch == "]":
                    if mystack and mystack[-1] == "[":
                        mystack.pop()
                    else:
                        return False

                elif ch == "}":
                    if mystack and mystack[-1] == "{":
                        mystack.pop()
                    else:
                        return False

        return False if mystack else True
```

```python
### https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/
### 1047. Remove All Adjacent Duplicates In String

class Solution:
    def removeDuplicates(self, s: str) -> str:
        mystack = []

        for ch in s:
            if mystack and ch == mystack[-1]:
                mystack.pop()
            else:
                mystack.append(ch)
        
        return "".join(mystack)
```

```python
### https://leetcode.com/problems/min-stack/description/
### 155. Min Stack

class MinStack:

    def __init__(self):
        self.stack_list = []
        self.minstack_list = []
        

    def push(self, value: int) -> None:
        self.stack_list.append(value)
        if not self.minstack_list:
            self.minstack_list.append(value)
        else:
            if value <= self.minstack_list[-1]:
                self.minstack_list.append(value)
        

    def pop(self) -> None:
        if self.stack_list[-1] == self.minstack_list[-1]:
            self.minstack_list.pop()
        self.stack_list.pop()

    def top(self) -> int:
        return self.stack_list[-1]
        

    def getMin(self) -> int:
        return self.minstack_list[-1]
        


# Your MinStack object will be instantiated and called as such:
# obj = MinStack()
# obj.push(value)
# obj.pop()
# param_3 = obj.top()
# param_4 = obj.getMin()
```

```python
### https://leetcode.com/problems/remove-duplicate-letters/
### 316. Remove Duplicate Letters
from collections import Counter

class Solution:
    def removeDuplicateLetters(self, s: str) -> str:
        mystack = []
        myset = set()
        count = Counter(s)

        for ch in s:
            count[ch] -= 1

            if ch in myset:
                continue

            while mystack and mystack[-1] > ch and count[mystack[-1]] > 0:
                val = mystack.pop() 
                myset.remove(val)

            mystack.append(ch)
            myset.add(ch)

        return "".join(mystack)
```

```python
### https://leetcode.com/problems/removing-stars-from-a-string/
### 2390. Removing Stars From a String
class Solution:
    def removeStars(self, s: str) -> str:
        mystack = []

        for ch in s:
            if ch == "*" and mystack:
                mystack.pop()
            else:
                mystack.append(ch)

        return "".join(mystack)
                
```

```python
### https://leetcode.com/problems/evaluate-reverse-polish-notation/
### 150. Evaluate Reverse Polish Notation

class Solution:
    def evalRPN(self, tokens: List[str]) -> int:

        operators_list = ["+", "-", "*", "/"]
        mystack = []

        for ch in tokens:
            if ch in operators_list:
                val_1 = mystack.pop()
                val_2 = mystack.pop()
                print("val_1 = ", val_1)
                print("val_2 = ", val_2)

                if ch == "+":
                    mystack.append(val_1 + val_2)
                elif ch == "-":
                    mystack.append(val_2 - val_1)
                elif ch == "*":
                    mystack.append(val_1 * val_2)
                else:
                    mystack.append(int(val_2 / val_1))
            else:
                mystack.append(int(ch))
    
        return mystack[0]
```

```python
### https://leetcode.com/problems/basic-calculator-ii/description/
### 227. Basic Calculator II

class Solution:
    def calculate(self, s: str) -> int:

        sign = "+"
        num = 0
        mystack = []

        for index,ch in enumerate(s):
            if ch.isdigit():
                num = num * 10 + int(ch)

            if not ch.isdigit() and not ch.isspace() or index == len(s)-1:
                if sign == "+":
                    mystack.append(num)
                elif sign == "-":
                    mystack.append(-num)
                elif sign == "*":
                    mystack.append(mystack.pop() * num)
                elif sign == "/":
                    val = int(mystack.pop() / num)
                    mystack.append(val)

                sign = ch
                num = 0

        return sum(mystack) 
```
