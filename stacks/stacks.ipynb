```python
### https://leetcode.com/problems/valid-parentheses/
### 20. Valid Parentheses

class Solution:
    def isValid(self, s: str) -> bool:
        mystack = []

        for ch in s:
            if ch == "(" or ch == "[" or ch == "{":
                mystack.append(ch)
            else:
                if ch == ")":
                    if mystack and mystack[-1] == "(":
                        mystack.pop()
                    else:
                        return False
                    
                elif ch == "]":
                    if mystack and mystack[-1] == "[":
                        mystack.pop()
                    else:
                        return False

                elif ch == "}":
                    if mystack and mystack[-1] == "{":
                        mystack.pop()
                    else:
                        return False

        return False if mystack else True
```

```python
### https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/
### 1047. Remove All Adjacent Duplicates In String

class Solution:
    def removeDuplicates(self, s: str) -> str:
        mystack = []

        for ch in s:
            if mystack and ch == mystack[-1]:
                mystack.pop()
            else:
                mystack.append(ch)
        
        return "".join(mystack)
```

```python
### https://leetcode.com/problems/remove-duplicate-letters/
### 316. Remove Duplicate Letters

from collections import Counter
class Solution:
    def removeDuplicateLetters(self, s: str) -> str:
        mystack = []
        freq_map = Counter(s)
        my_set = set()
        for ch in s:
            freq_map[ch] -=1
            if ch in my_set:
                continue

            while mystack and ch < mystack[-1] and freq_map[mystack[-1]] > 0:
                val = mystack.pop()
                my_set.remove(val)

            mystack.append(ch)
            my_set.add(ch)

        return "".join(mystack)
```

```python
### https://leetcode.com/problems/removing-stars-from-a-string/
### 2390. Removing Stars From a String
class Solution:
    def removeStars(self, s: str) -> str:
        mystack = []

        for ch in s:
            if ch == "*" and mystack:
                mystack.pop()
            else:
                mystack.append(ch)

        return "".join(mystack)
                
```

```python
### https://leetcode.com/problems/evaluate-reverse-polish-notation/
### 150. Evaluate Reverse Polish Notation

class Solution:
    def evalRPN(self, tokens: list[str]) -> int:

        mystack = []
        operand_list = {"+", "-", "/", "*"}
        for val in tokens:
            if val in operand_list:
                val_b = mystack.pop()
                val_a = mystack.pop()

                if val == "+":
                    mystack.append(val_a + val_b)
                elif val == "-":
                    mystack.append(val_a - val_b)
                elif val == "/":
                    mystack.append(int(val_a/val_b))
                else:
                    mystack.append(val_a * val_b)
            else:
                mystack.append(int(val))
        
        return mystack[0]
```

```python
### https://leetcode.com/problems/basic-calculator-ii/description/
### 227. Basic Calculator II

class Solution:
    def calculate(self, s: str) -> int:
        previous_operand = "+"
        num = 0
        mystack = []
        operand_list = {"+", "-", "*", "/"}

        for index, ch in enumerate(s):
            if ch.isdigit():
                num = num * 10 + int(ch)
            if ch in operand_list or index == len(s)-1:
                if previous_operand == "+":
                    mystack.append(num)
                elif previous_operand == "-":
                    mystack.append(-num)
                elif previous_operand == "*":
                    val = mystack.pop()
                    res = val * num
                    mystack.append(res)
                elif previous_operand == "/":
                    val = mystack.pop()
                    res = int(val/num)
                    mystack.append(res)

                num = 0
                previous_operand = ch

        return sum(mystack)
```

```python
### https://leetcode.com/problems/longest-valid-parentheses/description/
### 32. Longest Valid Parentheses
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        my_stack = [-1]
        max_length = 0

        for index, ch in enumerate(s):
            if ch == "(":
                my_stack.append(index)
            else:
                my_stack.pop() 
                if my_stack == []:
                    my_stack.append(index)
                else:
                    length = index - my_stack[-1]
                    max_length = max(max_length, length)
            
        return max_length
```

```python
### https://leetcode.com/problems/next-greater-element-i/description/
### 496. Next Greater Element I
class Solution:
    def nextGreaterElement(self, nums1: list[int], nums2: list[int]) -> list[int]:
        next_greater  = {}
        my_stack = []
        res = []
        for num in nums2:
            while my_stack and num > my_stack[-1]:
                smaller_val = my_stack.pop()
                next_greater[smaller_val] = num
            my_stack.append(num)
        
        for val in my_stack:
            next_greater[val] = -1

        for num in nums1:
            res.append(next_greater[num])
        
        return res
```

```python
### https://leetcode.com/problems/online-stock-span/description/
### 901. Online Stock Span

class StockSpanner:

    def __init__(self):
        self.stock_prices = []
        

    def next(self, price: int) -> int:
        current_span = 1

        while self.stock_prices and price >= self.stock_prices[-1][0]:
            prev_price, prev_span = self.stock_prices.pop()
            current_span += prev_span


        self.stock_prices.append((price, current_span))
    
        return current_span


# Your StockSpanner object will be instantiated and called as such:
# obj = StockSpanner()
# param_1 = obj.next(price)
```

```python
### https://leetcode.com/problems/132-pattern/description/
### 456. 132 Pattern
class Solution:
    def find132pattern(self, nums: list[int]) -> bool:
        
        second_element = float("-inf")
        my_stack = []
        for num in reversed(nums):

            if num < second_element:
                return True

            while my_stack and num >  my_stack[-1]:
                second_element = my_stack.pop()    
    
            my_stack.append(num)
        
        return False



```

```python
### https://leetcode.com/problems/largest-rectangle-in-histogram/description/
### 84. Largest Rectangle in Histogram
class Solution:
    def largestRectangleArea(self, heights: list[int]) -> int:
        
        my_stack = []
        max_area = 0

        for current_index, current_height in enumerate(heights):
            start_index = current_index

            while my_stack and current_height < my_stack[-1][-1]:
                past_index, past_height = my_stack.pop()
                area = (current_index - past_index) * past_height # width * height
                max_area = max(max_area, area)
                start_index = past_index

            my_stack.append((start_index, current_height))

        for index, height in my_stack:
            area = (len(heights) - index) * height
            max_area = max(max_area, area)
        

        return max_area
```

```python
### https://leetcode.com/problems/car-fleet/description/
### 853. Car Fleet
class Solution:
    def carFleet(self, target: int, position: list[int], speed: list[int]) -> int:
        fleet_count = 0
        cars = sorted(zip(position, speed), reverse=True)
        prev_fleet_time = 0
        for position, speed in cars:
            time = ((target - position) / speed)
            if time > prev_fleet_time:
                fleet_count += 1
                prev_fleet_time = time
        return fleet_count
```
