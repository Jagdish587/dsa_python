```python
### https://leetcode.com/problems/valid-parentheses/
### 20. Valid Parentheses

class Solution:
    def isValid(self, s: str) -> bool:
        close_to_open = {")": "(", "}" : "{", "]" : "["}
        my_stack = []

        for ch in s:
            if ch in close_to_open:
                if my_stack and my_stack[-1] == close_to_open[ch]:
                    my_stack.pop()
                else:
                    return False
            else:
                my_stack.append(ch)
    


        return False if my_stack else True
```

```python
### https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/description/
### 1047. Remove All Adjacent Duplicates In String

class Solution:
    def removeDuplicates(self, s: str) -> str:
        mystack = []

        for ch in s:
            if mystack and ch == mystack[-1]:
                mystack.pop()
            else:
                mystack.append(ch)
        
        return "".join(mystack)
```

```python
### https://leetcode.com/problems/remove-duplicate-letters/
### 316. Remove Duplicate Letters

from collections import Counter
class Solution:
    def removeDuplicateLetters(self, s: str) -> str:
        mystack = []
        freq_map = Counter(s)
        my_set = set()
        for ch in s:
            freq_map[ch] -=1
            if ch in my_set:
                continue

            while mystack and ch < mystack[-1] and freq_map[mystack[-1]] > 0:
                val = mystack.pop()
                my_set.remove(val)

            mystack.append(ch)
            my_set.add(ch)

        return "".join(mystack)
```

```python
### https://leetcode.com/problems/removing-stars-from-a-string/
### 2390. Removing Stars From a String
class Solution:
    def removeStars(self, s: str) -> str:
        mystack = []

        for ch in s:
            if ch == "*" and mystack:
                mystack.pop()
            else:
                mystack.append(ch)

        return "".join(mystack)
                
```

```python
### https://leetcode.com/problems/evaluate-reverse-polish-notation/
### 150. Evaluate Reverse Polish Notation

class Solution:
    def evalRPN(self, tokens: list[str]) -> int:

        mystack = []
        operand_list = {"+", "-", "/", "*"}
        for val in tokens:
            if val not in operand_list:
                mystack.append(int(val))
            else:
                val_b = mystack.pop()
                val_a = mystack.pop()

                if val == "+":
                    mystack.append(val_a + val_b)
                elif val == "-":
                    mystack.append(val_a - val_b)
                elif val == "/":
                    mystack.append(int(val_a/val_b))
                else:
                    mystack.append(val_a * val_b)
        
        return mystack[0]
```

```python
### https://leetcode.com/problems/basic-calculator-ii/description/
### 227. Basic Calculator II

class Solution:
    def calculate(self, s: str) -> int:
        previous_operand = "+"
        num = 0
        mystack = []
        operand_list = {"+", "-", "*", "/"}

        for index, ch in enumerate(s):
            if ch.isdigit():
                num = num * 10 + int(ch)
            if ch in operand_list or index == len(s)-1:
                if previous_operand == "+":
                    mystack.append(num)
                elif previous_operand == "-":
                    mystack.append(-num)
                elif previous_operand == "*":
                    val = mystack.pop()
                    res = val * num
                    mystack.append(res)
                elif previous_operand == "/":
                    val = mystack.pop()
                    res = int(val/num)
                    mystack.append(res)

                num = 0
                previous_operand = ch

        return sum(mystack)
```

```python
### https://leetcode.com/problems/longest-valid-parentheses/description/
### 32. Longest Valid Parentheses
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        my_stack = [-1]
        max_length = 0

        for index, ch in enumerate(s):
            if ch == "(":
                my_stack.append(index)
            else:
                my_stack.pop() 
                if my_stack == []:
                    my_stack.append(index)
                else:
                    length = index - my_stack[-1]
                    max_length = max(max_length, length)
            
        return max_length
```

```python
### https://leetcode.com/problems/car-fleet/description/
### 853. Car Fleet

# Intution : 
"""
The rule to remember
- If a car behind takes less than or equal to the time of the fleet ahead, it catches up and joins that fleet.
- If it takes more time, it cannot catch up before the target, so it forms a new fleet.
"""

class Solution:
    def carFleet(self, target: int, position: list[int], speed: list[int]) -> int:
        my_stack = []
        cars  = sorted(zip(position, speed), reverse=True)

        for position, speed in cars:
            # time = distance / speed
            current_time = (target - position)/speed
            if not my_stack or current_time > my_stack[-1]:
                my_stack.append(current_time)
        
        return len(my_stack)
```

```python
### https://leetcode.com/problems/simplify-path/description/
### 71. Simplify Path

class Solution:
    def simplifyPath(self, path: str) -> str:

        my_stack = []
        split_str = path.split("/")
        for str in split_str:
            if my_stack and str == "..":
                my_stack.pop()
            else:
                if str != "" and str != ".." and str != ".":
                    my_stack.append(str)

        res = "/" + "/".join(my_stack)
                
        return res
```
